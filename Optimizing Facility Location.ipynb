```python
from gurobipy import Model, GRB

# Given data
demand = [300, 400, 450, 250]  # Demand per quarter
cost = [20, 22, 24, 26]  # Cost per charger per quarter
storage_cost = 3.50  # Storage cost per charger per quarter
num_quarters = len(demand)

# Create a Gurobi model
model = Model("Inventory Optimization")

# Decision variables
order = model.addVars(num_quarters, vtype=GRB.INTEGER, name="order")  # Number of chargers to order
total_storage = model.addVars(num_quarters, vtype=GRB.INTEGER, name="storage")  # Chargers in storage

# Constraints
for t in range(num_quarters):
    if t == 0:
        model.addConstr(order[t] == demand[t] + total_storage[t], name=f"inventory_balance_{t}")
    else:
        model.addConstr(order[t] + total_storage[t-1] == demand[t] + total_storage[t], name=f"inventory_balance_{t}")

# Objective function (Minimize total cost)
total_cost = sum(order[t] * cost[t] for t in range(num_quarters)) + sum(total_storage[t] * storage_cost for t in range(num_quarters))
model.setObjective(total_cost, GRB.MINIMIZE)

# Optimize model
model.optimize()

# Print results
if model.status == GRB.OPTIMAL:
    print("Optimal solution found:")
    for t in range(num_quarters):
        print(f"Quarter {t+1}: Order {order[t].x}, Storage {total_storage[t].x}")
    print(f"Total cost: {model.objVal}")
else:
    print("No optimal solution found.")
```
