```python
from gurobipy import *
```

```python
# Create a new model
model = Model("Optimizing_Invt")
```

```python
# Decision variables and their coefficients in the objective function
projects = ['x1', 'x2', 'x3', 'x4', 'x5']
coefficients = [2.8, 2.6, 1.5, 1.75, 0.37009]
```

```python
# Create decision variables and store them in a dictionary
decision_vars = {project: model.addVar(name=project) for project in projects}
```

```python
# Set the objective function
model.setObjective(sum(coefficients[i] * decision_vars[projects[i]] for i in range(len(projects))), GRB.MAXIMIZE)
```

```python
# Add the total investment constraint
model.addConstr(sum(decision_vars[project] for project in projects) <= 10, "Total_Investment")
```

```python
# Add non-negativity constraints using a loop
for project in projects:
    model.addConstr(decision_vars[project] >= 0, f"Non_Negativity_{project}")
```

```python
# Optimize the model
model.optimize()
```

```python
# Print the results
if model.status == GRB.OPTIMAL:
    print(f"Optimal solution found:")
    for project in projects:
        print(f"Invest in {project}: ${decision_vars[project].X * 1000:.2f}")
    print(f"Total Yield: ${model.ObjVal * 1000:.2f}")
else:
    print("No optimal solution found.")
```
